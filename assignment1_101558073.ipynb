```python
"""
author: Yigit Alkoc
assignment: #1
"""

# step b: create 4 variables
gym_member = "Alex Alliton"   # str
preferred_weight_kg = 20.5    # float
highest_reps = 25             # int
membership_active = True      # bool

# step c: create a dictionary named workout_stats
# type: dictionary with tuple values
workout_stats = {
    "Alex": (30, 45, 20),
    "Jamie": (40, 35, 30),
    "Taylor": (50, 40, 45)
}

# step d: calculate total workout minutes using a loop and add to dictionary
friends_list = ["Alex", "Jamie", "Taylor"]

for name in friends_list:
    # getting the minutes tuple
    minutes = workout_stats[name]
    # calculating sum
    total = minutes[0] + minutes[1] + minutes[2]
    # adding to dictionary with new key
    workout_stats[name + "_Total"] = total

# step e: create a 2d nested list called workout_list
# type: list of lists
workout_list = [
    list(workout_stats["Alex"]),
    list(workout_stats["Jamie"]),
    list(workout_stats["Taylor"])
]

# step f: slice the workout_list
print("--- yoga and running (first two) ---")
# printing first two items for everyone
for row in workout_list:
    print(row[0:2])

print("\n--- weightlifting (last two friends) ---")
# printing last item for the last two people
last_two_friends = workout_list[1:]
for row in last_two_friends:
    print(row[2])

# step g: check if any friend's total >= 120
print("\n--- checking activity level ---")
for name in friends_list:
    total_min = workout_stats[name + "_Total"]
    if total_min >= 120:
        print(f"Great job staying active, {name}!")

# step h: user input to look up a friend
print("\n--- friend lookup ---")
friend_name = input("enter a friend's name: ")

# checking if name exists
if friend_name in workout_stats:
    stats = workout_stats[friend_name]
    total_val = workout_stats[friend_name + "_Total"]
    print(f"stats for {friend_name}: {stats}")
    print(f"total minutes: {total_val}")
else:
    print(f"Friend {friend_name} not found in the records.")

# step i: friend with highest and lowest total workout minutes
print("\n--- highest and lowest ---")

# finding the highest and lowest manually
highest_val = -1
highest_name = ""
lowest_val = 1000
lowest_name = ""

for name in friends_list:
    current_total = workout_stats[name + "_Total"]
    
    # check for max
    if current_total > highest_val:
        highest_val = current_total
        highest_name = name
        
    # check for min
    if current_total < lowest_val:
        lowest_val = current_total
        lowest_name = name

print(f"highest total: {highest_name} ({highest_val} mins)")
print(f"lowest total: {lowest_name} ({lowest_val} mins)")
```
